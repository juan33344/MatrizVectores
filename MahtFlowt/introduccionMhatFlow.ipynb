##### importar libreria

```python
import matplotlib.pyplot as plt
```

##### Grafico Linial

## simulando Datos

```python
plt.figure(figsize=(100,6))

#datos de paises

#Colombia
x = [2016, 2017, 2018, 2019, 2020, 2021] #Años
y = [42, 43,45,47,48,50] #población

#Argentina
x1 = [2016, 2017, 2018, 2019, 2020, 2021] #Años
y1 = [41, 48,49,51,52,55] #población

#Peru
x2 = [2016, 2017, 2018, 2019, 2020, 2021] #Años
y2 = [38, 39,40,42,52,61] #población
```

#### configurando la visualizacion de los datos

```python
plt.plot(x, y, marker='o', linestyle='--', color='g', label='Colombia')
plt.plot(x1, y1, marker='d', linestyle='-', color='r', label='Argentina')
plt.plot(x2, y2, marker='s', linestyle='-', color='b', label='Perú')

#agregar los nombres de los ejes y el titulo del grafico
plt.title('Población de Países en Millones')
plt.xlabel('Año')
plt.ylabel('Población (millones)')
plt.title('Años vs Poblaciones')
plt.legend(loc='lower right')

plt.yticks([41, 44, 48, 50 ])

#guardar el grafico
plt.savefig('grafico_poblacion.png')

#mostrar el grafico
plt.show()
```

# Utilizando la libreria SEABORN

#### Para Intalar: pip intall seaborn
importamos libreria

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

### Cargando un data set de ejemplo
Vamos a cargar un dataset predetiminado de "propinas"

```python
tips = sns.load_dataset("tips")
tips.head()
```

#### Grafico de dispercion (sccater plot)

```python
sns.scatterplot(data=tips, x="total_bill", y="tip", hue="day")
#mostrar el grafico 
plt.show()
```

### Graficos de barras (bar plot)
Graficos de barras para la cantidad promedios de las propinas según el dia de la semana

```python
sns.barplot(x="day", y="tip", data=tips )
```

## Histograma 
Histograma de la columna total_bill

```python
sns.histplot(tips["total_bill"], kde=True)
```

## Grafico de Cajas 


```python
sns.boxplot(x="day", y="total_bill",hue="day", data=tips )
```

## Personalizacion de Grafico 
Grafico de dispercion con diferensacion por genero

```python
sns.scatterplot( x="total_bill", y="tip", hue="sex", style="day", data=tips)
```

## Grafico de Modelacion con Mapa de calor (heatmat)

```python
# Seleccionar solo los columnas numericas de DataSet
column_numericas = tips.select_dtypes(include=["float64", "int64"])

#Calcular la Matriz de Correlación de columnas numericas
correlacion = column_numericas.corr()

# crear el mapa de calor
sns.heatmap(correlacion, annot=True, cmap="coolwarm", fmt=".2f")
```

## Grafico de Lineas
Crear Grafico de linias para total_bill vs tip

```python
sns.lineplot(x="total_bill", y="tip", hue="day", data=tips)
```

## Grafico de Violin (violin plot)
Grafico de violin de total_bill por dia 

```python
sns.violinplot(x="day", y="total_bill", hue="sex", data=tips, split=True)
```

## Grafico de Pares (pair plot)


```python
sns.pairplot(tips)
```

## Grafico de Conteo (cout plot)
Crear un Grafico de Conteo de fumadores vs no fumadores

```python
sns.countplot(x="smoker", data=tips )
```

## Grafico de mapa de puntos (strip plot)
Grafico de mapa de puntos para total_bill segun el dia


```python
sns.stripplot(x="day", y="total_bill", hue="sex", data=tips)
```

## Grafico de Puntos cotegoria (point plot)
Grafico de puntos para tip según el dia y el género 

```python
sns.pointplot(x="day", y="tip", hue="sex", data=tips)
```

## Grafico de distribucion con junta (join plot)
Crear un grafico de distribucion conjunta para total_bill y tip.

```python
sns.jointplot(x="total_bill", y="tip", data=tips, kind="scatter")
```

## Grafico de Kde (kde plot)
Crear grafico de kde para el total_bill

```python
sns.kdeplot(tips["total_bill"], fill=True)
```

## Grafico de Matriz de Relaciones Categoricas (cat plot)
Este Grafico es util para mostrar multiples convinaciones de variables categoricas y numericas
crear un catplot para propinas segun dia y género

```python
sns.catplot(x="day", y="tip", hue="sex",  kind="bar", data=tips)
```

## Grafico de FacetGrid, util para separar categorias y multiples grafico en una cuadricula.
Crear un FecetGrid para total_bill vs tip, separar por el fumador 

```python
g = sns.FacetGrid(tips, col="smoker")
g.map(sns.scatterplot, "total_bill", "tip")
```
